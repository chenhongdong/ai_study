# 记忆（Memory）
Agemt的记忆分两类：
- 短期记忆: 当前任务或会话的上下文
- 长期记忆: 跨任务或会话的经验与知识

## 添加记忆
AgentState实现了LangChain短期记忆，而会话历史就是AgentState的一部分。

LangChain提供了Checkpointer对象来保存AgentState，每一次用户与AI交互都会生成一个快照，记录为一个checkpoint。
同一会话的多个checkpoint为一组，用同一个thread_id来标记。

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
  'deepseek-chat',
  checkpointer=InMemorySaver()
)
```

```python
from langchain.messages import HumanMessage

config = {'configurable': {'thread_id': 'thread_1'}}

response = agent.invoke(
  {'messages': HumanMessage('我家的小白猫11岁了')},
  config
)

print(response)
```

```python
response = agent.invoke(
  {'messages': HumanMessage('我养的猫叫什么名字？')},
  config
)

print(response)
```

```python
response = agent.stream(
  {'messages': [HumanMessage('我的小猫名字就叫小白，大名叫陈百万')]},
  config,
  stream_mode='messages'
)

for chunk,metadata in response:
  print(chunk.content, end='', flush=True)
```

## Memory持久化存储
这里我们选择使用SqLite作为存储方案，首先安装依赖：`uv add langgraph-chenkpoint-sqlite`

接着按照步骤使用：
1. 导入依赖
2. 初始化checkpointer
3. 自动建表
4. 创建agent，指定checkpointer

```python
import sqlite3
from langgraph.checkpoint.sqlite import SqliteSaver

# 连接sqlite
connection = sqlite3.connect('resources/checkpoint.db', check_same_thread=False)
# 初始化checkpointer
checkpointer = SqliteSaver(connection)
# 自动建表
checkpointer.setup()

# 创建agent
agent = create_agent(
  'deepseek-chat',
  checkpointer=checkpointer
)

```

```python
from langchain.messages import HumanMessage

config = {'configurable': {'thread_id': 'thread_2'}}

response = agent.stream(
  {'messages': [HumanMessage('我想养一只柴犬黄色的')]},
  config,
  stream_mode='messages'
)

for chunk,metadata in response:
  print(chunk.content, end='', flush=True)
```

```python
response = agent.stream(
  {'messages': [HumanMessage('我刚才说想养猫还是狗来着？')]},
  config,
  stream_mode='messages'
)

for chunk,metadata in response:
  print(chunk.content, end='', flush=True)
```

## 记忆管理
当会话历史过长时，可能会超出模型的上下文窗口限制，常见的解决方法有：
- 修剪消息和删除消息
- 总结摘要消息(推荐)

```python
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver

# 初始化中间件
middleware = SummarizationMiddleware(
    model='deepseek-chat',
    trigger=('messages', 3),  # 触发时机，当消息数量超过3条时，进行摘要总结
    keep=('messages', 1)  # 保留的会话数，超过2条
  )

checkpointer = InMemorySaver()

agent = create_agent(
  'deepseek-chat',
  checkpointer=checkpointer,
  middleware=[middleware]
)
config = {'configurable': {'thread_id': 'thread_3'}}

# 多轮对话
agent.invoke({'messages': HumanMessage('你觉得未来值得投入的行业是AI吗？')}, config)
agent.invoke({'messages': HumanMessage('2027年还会有前端岗位吗？')}, config)
agent.invoke({'messages': HumanMessage('AI会在2027年出现泡沫吗？')}, config)

final = agent.invoke({'messages': HumanMessage('如果泡沫出现了普通人该怎么办？在泡沫中又有什么机会呢')}, config)
```

```python
print(final)
```
